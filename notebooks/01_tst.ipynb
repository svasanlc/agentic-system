```python
import os
openai_api_key = os.getenv('OPENAI_API_KEY')

if openai_api_key:
    print(f"OPENAI_API_KEY exists and begins {openai_api_key[:8]}")
else:
    print("OPENAI_API_KEY is not set")

```

```python
from openai import OpenAI
```

```python
# Now I will create an instance of the OpenAI class
openai = OpenAI()
```

```python
messages = [{"role": "user", "content": "Tell me a fun fact"}]
```

```python
messages
```

```python
# This uses GPT 5.4 nano, which is a cheap model
response = openai.chat.completions.create(model="gpt-5.4-nano", messages=messages)
print(response.choices[0].message.content)
```

```python
# now lets ask a question
question = "Please propose a hard, challenging question to assess someone's IQ. Respond only with the question."
messages = [{"role": "user", "content": question}]
```

```python
#get the response for the question you asked above and print it.
response = openai.chat.completions.create(model="gpt-5.4-nano", messages=messages)
question = response.choices[0].message.content
print(question)
```

```python
#Form a new messages list of the question you created above
messages = [{"role": "user", "content": question}]
messages
```

```python
# Now ask the model

response = openai.chat.completions.create(model="gpt-5.4-mini", messages=messages)
answer = response.choices[0].message.content
print(answer)
```

```python
#format the same response you got above in a readable format
from IPython.display import Markdown, display

display(Markdown(answer))
```

```python
message = f"""
Here is a question:
{question}

And here is a possible answer that might be correct or incorrect:
{answer}

Please evaluate if the answer is correct or incorrect.
"""

print(message)
```

```python
messages = [{"role": "user", "content": message}]
response = openai.chat.completions.create(model="gpt-5.4", messages=messages)
print(response.choices[0].message.content)
```
