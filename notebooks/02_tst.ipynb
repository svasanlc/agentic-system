```python
# This test is to test the OpenAI API by using the API to call LLMs
# start with imports
import os
import json
from dotenv import load_dotenv
from openai import OpenAI
from IPython.display import Markdown, display
```

```python
# Always remember to load_dotenv()
load_dotenv(override=True)
```

```python
openai_api_key = os.getenv("OPENAI_API_KEY")

if openai_api_key:
    print(f"OPENAI_API_KEY exists and begins with {openai_api_key[:8]}")
else:
    print("OPENAI_API_KEY does not exist")
```

```python
# we want the LLM to come up with a question, so we want to word the our request as shown and put that in a message. 
# we are printing the message to see what we are going to ask the LLM.
request = """
Please come up with a challenging, nuanced question with a succint answer, that I can ask a nmber of LLMs to evaluate their intelligence.
Not a mathematical puzzle, but more of a thought-provoking question which requires intelligent insight.
Include in your question that the answer must be short.
"""
request += "Answer only with the question, no explanation."
messages = [{"role": "system", "content": request}]
messages
```

```python
# Now we want to create an instance of the OpenAI class and use the chat.completions.create method to get the response from the LLM.
# in other words, we want the LLM to come up with a question.
openai = OpenAI()
response = openai.chat.completions.create(model="gpt-4o-mini", messages=messages)
question = response.choices[0].message.content
display(Markdown(question))

```

```python
competitors = []
answers = []
messages = [{"role": "user", "content": question}]
```

```python
def record(model_name, answer):
    competitors.append(model_name)
    answers.append(answer)
    display(Markdown(answer))
```

```python
# Reasoning effort can be none, low, medium, high, or xhigh
model_name = "gpt-5.4-nano"

response = openai.chat.completions.create(model=model_name, messages=messages, reasoning_effort="none")
answer = response.choices[0].message.content

record(model_name, answer)
```
